```python
## Create clean measles vaccination outcome
```

```python
nics["measles_vaccinated"] = nics["IM16"].map({
    1.0: 1,
    2.0: 0
})
```

```python
nics["measles_vaccinated"].value_counts(dropna=False)
```

```python
nics.groupby("source")["measles_vaccinated"].value_counts(
    dropna=False
).sort_index()
```

```python
## IM16 Availability
```

```python
im16_availability = (
    nics.groupby("source")["measles_vaccinated"]
    .agg(
        valid="count",
        total="size"
    )
)

im16_availability["missing"] = (
    im16_availability["total"] - im16_availability["valid"]
)

im16_availability["valid_percent"] = (
    im16_availability["valid"] /
    im16_availability["total"] * 100
)

im16_availability
```

```python
im16_availability = (
    nics.groupby("source")["measles_vaccinated"]
    .agg(
        valid="count",
        total="size"
    )
)

im16_availability["missing"] = (
    im16_availability["total"] - im16_availability["valid"]
)

im16_availability["valid_percent"] = (
    im16_availability["valid"] /
    im16_availability["total"] * 100
)

im16_availability
```

```python
## IM16 Availability

#IM16 was available for 2,058 of the 6,268 children in the analytical dataset (32.8%). 
#Availability was 32.1% among core MICS records and 38.6% among supplemental NICS records.

#Records with missing IM16 were also found to have missing values for other immunisation 
#variables (IM7, IM8, and IM15A). Therefore, missing IM16 values were not interpreted as 
#measles-specific nonresponse and were not recoded as unvaccinated.

#For analyses of measles vaccination status, children with valid IM16 responses 
#(IM16 = 1 or 2) will form the outcome-specific analytical denominator. IM16 codes 8 
#and 9 are treated as indeterminate.
```
